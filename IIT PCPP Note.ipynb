### Sets

```python
my_set={1,2,3}
my_set.add(4) #add single element
print(my_set)
```

```python
my_set={1,2,3}
my_set.update([5,6]) #add more elements
print(my_set)
```

```python
my_set={1,2,3}
my_set.remove(2) #remove an element
my_set.discard(10) #remove an element without an error that might not in the set
my_set.pop() #remove a random element

print(my_set)
```

```python
my_set={1,2,3}
my_set.clear() #remove all elements
print(my_set)
```

```python
my_set3={101,102,103,104,105,106,107,108}
my_set3.add(109)
my_set3.update([110,111,112])
my_set3.remove(101)
my_set3.discard(200)
my_set3.pop()
print(my_set3)
```

```python
my_set3={101,102,103,104,105,106,107,108}
my_set3.add(109)
my_set3.update([110,111,112])
my_set3.remove(101)
my_set3.discard(200)
my_set3.pop()
my_set3.clear()
print(my_set3)
```
